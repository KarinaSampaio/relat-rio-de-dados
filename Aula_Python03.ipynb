# If / Else

If = testa a condição e apresenta a condição

Else: representa o caminho alternativo; O else, atende caso nenhuma condição anterior esteja verdadeira.

```python
nota = 6
if nota >= 7:
  print('Aprovado')
else:
  print('Reprovado')
```

# Elif

Usa quando existem mais de duas possibilidades; O elif significa "senão, se", ou seja: Ele oferece outra condição quuandos as anteriores não foram atendidas.

```python
nota = 8.9
if nota >= 7:
  print('Aprovado')
elif nota >= 5:
  print('Recuperação')
else:
  print('Reprovado')
```

# Criando uma função para classificar cada aluno em: média e frequencia

```python
def classificar(media, frequencia):
  if media >= 7 and frequencia >= 75:
    return 'Aprovado'
  elif media >= 5 and frequencia >= 75:
    return 'Recuperação'
  else:
    return 'Reprovado'

df['media'] = (df['nota_1'] + df['nota_2']) / 2
df['situação'] = df.apply(lambda linha: classificar(linha['media'], linha['frequencia']), axis=1)
df.head()
```

# Desafio loja

1. Fazer o upload de pedidos_loja.csv e criar uma regra de desconto:



*   Cliente VIP recebe 15%
*   compra acima de R$ 150 recebe 10%
*   quem usou cupom recebe 5%
*   demais casos não recebem desconto.





```python
uploaded = files.upload()
arquivo = next(iter(uploaded))
import pandas as pd
pedidos = pd.read_csv(arquivo)
```

```python
df = pd.read_csv(arquivo)
df.head()
```

```python
# Desafio

def desconto(valor, tipo_cliente, usou_cupom):
  if tipo_cliente == 'vip':
    return 0.15
  elif valor > 150:
    return 0.10
  elif usou_cupom == True:
    return 0.05
  else:
    return 0.00

pedidos['percentual_desconto'] = pedidos.apply(lambda linha: desconto(
    linha['valor_compra'],
    linha['tipo_cliente'],
    linha['usou_cupom']
),
axis=1
)

#Calcular o valor final para as duas casas decimais
pedidos['valor_final'] = (pedidos['valor_compra'] * (1 - pedidos['percentual_desconto'])).round(2)
pedidos.head()
```

```python
# Sem lambda recebe uma linha e devolve o percentual de desconto
def calcular_desconto(linha):
  if linha['tipo_cliente'] == 'vip':
    return 0.15

  elif linha['valor_compra'] > 150:
    return 0.10

  elif linha['usou_cupom']:
    return 0.05

  else:
    return 0.00

# Aplica a função de cada linha da tabela
pedidos['percentual_desconto'] = pedidos.apply(calcular_desconto, axis=1).round(2)
pedidos.head()
```
