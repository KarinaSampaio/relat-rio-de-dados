## Lógica Booleana, operadores e condicionais

Tomamndo decisões de acordo com condições

# 1. Booleanos

Booleano representa apenas dois estados: True ou False

```python
maior_de_idade = True
possui_cupom = False
print(maior_de_idade)
print(possui_cupom)
```

# 2. Operadores Relacionais

Eles comparam valores e retornam True ou False

```python
nota = 7.5
frequencia = 82
print(nota >= 7)
print(frequencia >= 75)
print(nota < 5)
print(frequencia != 0)
```

# 3. Operadores Lógicos

"and" exige que as duas condições sejam verdadeiras. "or" exige pelo menos uma verdadeira. "not" inverter o resultado.

```python
nota = 6
frequencia = 80
aprovado = nota >= 7 and frequencia >= 75
print(aprovado)
```

## Exemplo com clientes churn

```python
plano_ativo = True
dias_sem_acesso = 2
quantidade_reclamacoes = 1
pagamento_em_atraso = False

churn_confirmado = not plano_ativo

risco_churn = ( plano_ativo and dias_sem_acesso >= 30 and (quantidade_reclamacoes >= 2 or pagamento_em_atraso))

if churn_confirmado:
   print('Cliente em churn: plano cancelado.')

elif risco_churn:
  print('Cliente com alto risco de churn.')

else:
  print('Cliente ativo e engajado')
```
