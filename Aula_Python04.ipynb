# 1. Listas

Guardam vários valores em uma unica variável.

```python
notas = [7.5, 8.0, 6.0, 9.0, 5.5, 7.9, 10.0, 8.1]
print(notas)
print(notas[7])
```

# 2. for

percorre uma sequência de valores

```python
nomes = ['Ana', 'Bruno', 'Carla', 'Tiago', 'Fabio']
for nome in nomes:
  print('Olá', nome)
```

# 3. range

range() cria uma sequência numérica.

```python
for numero in range(1, 11, 2):
 print(numero)
```

```python
for numero in range(1, 6):
  print(numero)
```

# 4. Contador

aumenta quando algo acontece

```python
notas = [7.5, 8.0, 6.0, 9.0, 5.5]
aprovados = 0
for nota in notas:
  if nota >= 8:
    aprovados += 1
print('Aprovados:', aprovados)
```

```python
notas = [7.5, 8.0, 6.0, 9.0, 5.5, 6.2, 9.0]
aprovados = 0
reprovados = 0

for nota in notas:
  if nota >= 7:
    aprovados += 1
  else:
    reprovados +=1

print('Aprovados:', aprovados)
print('Reprovados:', reprovados)
```

# 5. Acumulador

Soma valores aos poucos.

```python
vendas = [1200, 2000, 600, 3500, 1800]
total = 0
for venda in vendas:
  total += venda
print('Total:', total)
```

# 6. While

repete enquanto uma condição for verdadeira

```python
contador = 1
while contador <= 10:
  print(contador)
  contador += 1
```

# Quando usar For e While

1. Usamos o for quando existe lista ou quantidade conhecida
2. Usamos o while quando temos uma condição
3. for vai percorrer notas
4. while ele tenta acertar a senha

# Base de vendas

vamos produzir um relatório simples fazendo o upload do arquivo vendas_semana.csv

```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
arquivo = next(iter(uploaded))
df = pd.read_csv(arquivo)
df.head()
```

# Projeto integrador: relatório

Crie um relatório com:

1. total de vendas;
2. média diária;
3. maior venda;
4. dias que bateram a meta;
5. uma conclusão em linguagem de negócio.

# Conferindo quantidade de registros e a existência de valores ausentes

```python
print('Quantidade de dias registrados:', len(df))

print('Valores ausentes por coluna:')
print(df.isna().sum())
```

## Calculando o total de vendas

quanto foi vendido durante a semana?

```python
# Soma os valores de vendas dos sete dias
total = df['vendas'].sum()

print(f'Total de vendas: R$ {total: .2f}')
```

## Calculando a média diária

quanto a loja vendeu, em media, por dia?

```python
# Calcula a média dos valores diários
media = df['vendas'].mean()

print(f'Média diária: R$ {media: .2f}')
```

## Identificando a maior venda diária

qual foi o maior valor vendido em um único dia?

```python
maior = df['vendas'].max()

print(f'Maior venda diária: R$ {maior: .2f}')
```

## Contagem de dias que atingiram a meta

em quantos dias as vendas foram maiores ou iguais a meta?

```python
dias_meta = 0

for indice, linha in df.iterrows():
  if linha['vendas'] >= linha['meta']:
    dias_meta += 1

quantidade_dias = len(df)

print(
    f'Dias que bateram a meta: '
    f'{dias_meta} de {quantidade_dias}'
    )
```

## Verificar se a meta total da semana foi atingida

o valor total vendido alcançou a soma das metas?

```python
meta_semana = df['meta'].sum()

diferenca = total - meta_semana

print(f'Meta semanal: R$ {meta_semana: .2f}')
print(f'Diferença entre vendas e meta: R$ {diferenca: .2f}')
```
